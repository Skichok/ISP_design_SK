```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
```

```python
class LeNet(nn.Module):
    def __init__(self):
        super(LeNet, self).__init__()
        self.conv1 = nn.Conv2d(1, 6, kernel_size=5)
        self.conv2 = nn.Conv2d(6, 16, kernel_size=5)
        self.fc1 = nn.Linear(16 * 4 * 4, 120)  # 4x4 - размер после сверток
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, x):
        x = torch.relu(self.conv1(x))
        x = torch.max_pool2d(x, 2)
        x = torch.relu(self.conv2(x))
        x = torch.max_pool2d(x, 2)
        x = x.view(-1, 16 * 4 * 4)
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

```python
# Трансформации для нормализации данных
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))  # Нормализация для черно-белых изображений
])

# Загрузка набора данных MNIST
train_dataset = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_dataset = datasets.MNIST(root='./data', train=False, transform=transform, download=True)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = LeNet().to(device)

# Функция потерь и оптимизатор
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
# Функция для тренировки модели
def train(model, train_loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
    return running_loss / len(train_loader.dataset)

# Функция для тестирования модели
def test(model, test_loader, criterion, device):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    return 100 * correct / total
```

```python
num_epochs = 10
for epoch in range(num_epochs):
    train_loss = train(model, train_loader, criterion, optimizer, device)
    accuracy = test(model, test_loader, criterion, device)
    print(f'Epoch {epoch + 1}/{num_epochs}, Loss: {train_loss:.4f}, Accuracy: {accuracy:.2f}%')
```

```python
from torchvision import datasets, transforms
from PIL import Image

# Задаем трансформацию для преобразования изображений в тензоры
transform = transforms.ToTensor()

# Загружаем датасет MNIST
mnist_dataset = datasets.MNIST(root='./data', train=True, transform=transform, download=True)

# Получаем изображение и метку по индексу
image, label = mnist_dataset[10]

# Преобразуем тензор обратно в формат, понятный Pillow
image = transforms.ToPILImage()(image)

# Отображаем изображение с помощью PIL
image.show(title=f"Label: {label}")
print(image)
```

```python
modeli = torch.load('lenet_mnist_full.pth')  # Загружаем всю модель
modeli.eval()  # Устанавливаем модель в режим инференса
```

```python
# Функция для подготовки изображения
def prepare_image(image_path):
    # Открываем изображение
    image = Image.open(image_path).convert('L')  # Конвертируем в оттенки серого
    image = image.resize((28, 28))  # Приводим к размеру 28x28
    
    # Применяем трансформации
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))  # Нормализация
    ])
    
    image = transform(image)
    image = image.unsqueeze(0)  # Добавляем размерность для батча
    return image
```

```python
# Путь к вашему изображению BMP
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
image_path = './data/1.bmp'  # Укажите путь к вашему BMP изображению
image_tensor = prepare_image(image_path)
image_tensor = image_tensor.to(device)
# Инференс
with torch.no_grad():  # Отключаем градиенты
    output = modeli(image_tensor)
    _, predicted = torch.max(output, 1)  # Получаем индекс класса с максимальным значением
    print(f'Predicted Label: {predicted.item()}')  # Выводим предсказанную метку
```
